```python
def accept_marks(num_subjects=5):
    """Loop to accept valid marks (0-100) for a given number of subjects."""
    marks = []
    print(f"\n--- Enter Marks for {num_subjects} Subjects ---")
    for i in range(1, num_subjects + 1):
        while True:
            try:
                mark = float(input(f"Enter marks for Subject {i} (0-100): "))
                if 0 <= mark <= 100:
                    marks.append(mark)
                    break
                else:
                    print("Invalid input! Marks must be between 0 and 100.")
            except ValueError:
                print("Invalid input! Please enter a numerical value.")
    return marks


def calculate_total(marks):
    """Calculate the total sum of marks."""
    total = 0.0
    for mark in marks:
        total += mark
    return total


def calculate_percentage(total, num_subjects=5):
    """Calculate percentage based on total marks."""
    return total / num_subjects


def assign_grade(percentage):
    """Assign a letter grade based on percentage."""
    if percentage >= 90:
        return 'A+'
    elif percentage >= 80:
        return 'A'
    elif percentage >= 70:
        return 'B'
    elif percentage >= 60:
        return 'C'
    elif percentage >= 50:
        return 'D'
    else:
        return 'F'


def determine_pass_fail(marks, passing_mark=40):
    """Determine pass/fail status (requires passing every subject)."""
    for mark in marks:
        if mark < passing_mark:
            return "Fail"
    return "Pass"


def display_result(marks, total, percentage, grade, status):
    """Display the detailed student result summary."""
    print("\n" + "=" * 30)
    print("      STUDENT RESULT REPORT      ")
    print("=" * 30)
    for idx, mark in enumerate(marks, start=1):
        print(f"Subject {idx}: {mark:.2f}")
    print("-" * 30)
    print(f"Total Marks : {total:.2f}")
    print(f"Percentage  : {percentage:.2f}%")
    print(f"Grade       : {grade}")
    print(f"Status      : {status}")
    print("=" * 30)


def main():
    """Main function to run the Student Result Management System."""
    num_subjects = 5
    marks = accept_marks(num_subjects)
    total = calculate_total(marks)
    percentage = calculate_percentage(total, num_subjects)
    grade = assign_grade(percentage)
    status = determine_pass_fail(marks)

    display_result(marks, total, percentage, grade, status)


if __name__ == "__main__":
    main()
```

```python
def check_balance(balance):
    """Display the current account balance."""
    print(f"\nCurrent Balance: ${balance:.2f}")


def deposit(balance, transactions):
    """Deposit money into the account."""
    try:
        amount = float(input("\nEnter amount to deposit: $"))
        if amount > 0:
            balance += amount
            transactions.append(f"Deposited: ${amount:.2f}")
            print(f"Successfully deposited ${amount:.2f}")
        else:
            print("Deposit amount must be greater than 0.")
    except ValueError:
        print("Invalid input! Please enter a numerical amount.")
    return balance


def withdraw(balance, transactions):
    """Withdraw money from the account if sufficient funds exist."""
    try:
        amount = float(input("\nEnter amount to withdraw: $"))
        if amount <= 0:
            print("Withdrawal amount must be greater than 0.")
        elif amount > balance:
            print("Insufficient funds! Transaction cancelled.")
        else:
            balance -= amount
            transactions.append(f"Withdrew: ${amount:.2f}")
            print(f"Successfully withdrew ${amount:.2f}")
    except ValueError:
        print("Invalid input! Please enter a numerical amount.")
    return balance


def display_transactions(transactions):
    """Display all recorded transactions."""
    print("\n--- Transaction History ---")
    if not transactions:
        print("No transactions recorded yet.")
    else:
        for idx, entry in enumerate(transactions, start=1):
            print(f"{idx}. {entry}")


def main():
    """Main application loop with menu options."""
    balance = 0.0
    transactions = []

    while True:
        print("\n==========================")
        print("   BANKING APPLICATION    ")
        print("==========================")
        print("1. Check Balance")
        print("2. Deposit")
        print("3. Withdraw")
        print("4. Transaction History")
        print("5. Exit")
        
        choice = input("Select an option (1-5): ").strip()

        if choice == '1':
            check_balance(balance)
        elif choice == '2':
            balance = deposit(balance, transactions)
        elif choice == '3':
            balance = withdraw(balance, transactions)
        elif choice == '4':
            display_transactions(transactions)
        elif choice == '5':
            print("\nThank you for using our Banking Application. Goodbye!")
            break
        else:
            print("Invalid selection! Please choose an option from 1 to 5.")


if __name__ == "__main__":
    main()
```

```python
# 3. Inventory Management

def add_product(inventory):
    """Add a new product or increment stock if it already exists."""
    name = input("\nEnter product name: ").strip().title()
    if not name:
        print("Product name cannot be empty.")
        return

    # Check if product already exists
    for product in inventory:
        if product['name'].lower() == name.lower():
            print(f"'{name}' already exists! Use 'Update Quantity' to change stock.")
            return

    try:
        price = float(input("Enter product price ($): "))
        quantity = int(input("Enter product quantity: "))

        if price < 0 or quantity < 0:
            print("Price and quantity must be non-negative values.")
            return

        inventory.append({
            "name": name,
            "price": price,
            "quantity": quantity
        })
        print(f"Product '{name}' added successfully!")
    except ValueError:
        print("Invalid input! Price must be a number and quantity must be an integer.")


def display_products(inventory):
    """Display all products in a structured format."""
    print("\n" + "=" * 45)
    print("             CURRENT INVENTORY             ")
    print("=" * 45)
    if not inventory:
        print("No products available in inventory.")
    else:
        print(f"{'Name':<20} {'Price ($)':<12} {'Quantity':<10}")
        print("-" * 45)
        for prod in inventory:
            print(f"{prod['name']:<20} {prod['price']:<12.2f} {prod['quantity']:<10}")
    print("=" * 45)


def search_product(inventory):
    """Search for a product by name (case-insensitive)."""
    search_name = input("\nEnter product name to search: ").strip().lower()
    found = False
    
    for prod in inventory:
        if prod['name'].lower() == search_name:
            print("\nProduct Found:")
            print(f"Name    : {prod['name']}")
            print(f"Price   : ${prod['price']:.2f}")
            print(f"Quantity: {prod['quantity']}")
            found = True
            break

    if not found:
        print(f"Product matching '{search_name}' was not found.")


def update_quantity(inventory):
    """Update the stock quantity for an existing product."""
    search_name = input("\nEnter product name to update quantity: ").strip().lower()

    for prod in inventory:
        if prod['name'].lower() == search_name:
            try:
                new_qty = int(input(f"Enter new quantity for '{prod['name']}': "))
                if new_qty < 0:
                    print("Quantity cannot be negative.")
                    return
                prod['quantity'] = new_qty
                print(f"Quantity for '{prod['name']}' updated to {new_qty}.")
                return
            except ValueError:
                print("Invalid input! Quantity must be an integer.")
                return

    print(f"Product matching '{search_name}' was not found.")


def calculate_total_inventory_value(inventory):
    """Calculate and display the overall financial value of the inventory."""
    total_value = 0.0
    for prod in inventory:
        total_value += prod['price'] * prod['quantity']
    
    print(f"\nTotal Inventory Value: ${total_value:.2f}")
    return total_value


def main():
    """Main application loop."""
    inventory = []

    while True:
        print("\n=================================")
        print("   INVENTORY MANAGEMENT SYSTEM   ")
        print("=================================")
        print("1. Add Product")
        print("2. Display All Products")
        print("3. Search Product")
        print("4. Update Product Quantity")
        print("5. Calculate Total Inventory Value")
        print("6. Exit")

        choice = input("Select an option (1-6): ").strip()

        if choice == '1':
            add_product(inventory)
        elif choice == '2':
            display_products(inventory)
        elif choice == '3':
            search_product(inventory)
        elif choice == '4':
            update_quantity(inventory)
        elif choice == '5':
            calculate_total_inventory_value(inventory)
        elif choice == '6':
            print("\nExiting Inventory Management System. Goodbye!")
            break
        else:
            print("Invalid option! Please enter a number between 1 and 6.")


if __name__ == "__main__":
    main()
```

```python
# 5. Number Analysis Tool

def analyze_numbers(numbers):
    """Analyzes a list of numbers without using min(), max(), or sum().

    Returns a dictionary with all statistical properties.
    """
    if not numbers:
        return None  # Return None if the list is empty

    # Initialize tracking variables using the first element for min/max
    largest = numbers[0]
    smallest = numbers[0]
    total = 0
    even_count = 0
    odd_count = 0
    positive_count = 0
    negative_count = 0

    # Loop through each number in the list
    for num in numbers:
        # 1. Total calculation
        total += num

        # 2. Largest and Smallest checks
        if num > largest:
            largest = num
        if num < smallest:
            smallest = num

        # 3. Even and Odd checks (integers only)
        if isinstance(num, int):
            if num % 2 == 0:
                even_count += 1
            else:
                odd_count += 1

        # 4. Positive and Negative checks
        if num > 0:
            positive_count += 1
        elif num < 0:
            negative_count += 1

    # Average calculation
    average = total / len(numbers)

    # Return results in a structured dictionary
    return {
        "largest": largest,
        "smallest": smallest,
        "total": total,
        "average": average,
        "even_count": even_count,
        "odd_count": odd_count,
        "positive_count": positive_count,
        "negative_count": negative_count,
    }


def main():
    """Main function to demonstrate the Number Analysis Tool."""
    # Sample input list (can include positive, negative, odd, and even integers)
    sample_list = [12, -7, 5, 0, 18, -3, 22, -10, 9]

    results = analyze_numbers(sample_list)

    if results:
        print("Input List:", sample_list)
        print("\n" + "=" * 30)
        print("     NUMBER ANALYSIS REPORT    ")
        print("=" * 30)
        print(f"Largest Number : {results['largest']}")
        print(f"Smallest Number: {results['smallest']}")
        print(f"Total Sum      : {results['total']}")
        print(f"Average        : {results['average']:.2f}")
        print(f"Even Count     : {results['even_count']}")
        print(f"Odd Count      : {results['odd_count']}")
        print(f"Positive Count : {results['positive_count']}")
        print(f"Negative Count : {results['negative_count']}")
        print("=" * 30)
    else:
        print("The list provided is empty.")


if __name__ == "__main__":
    main()
```

```python
# 6. Employee Salary Analyzer

def calculate_total_payroll(salaries):
    """Calculate the sum of all employee salaries."""
    total = 0.0
    for salary in salaries.values():
        total += salary
    return total


def calculate_average_salary(total_payroll, num_employees):
    """Calculate the average salary across all employees."""
    if num_employees == 0:
        return 0.0
    return total_payroll / num_employees


def find_highest_salary(salaries):
    """Find the highest earning employee(s) and their salary."""
    if not salaries:
        return None, 0.0

    max_salary = float('-inf')
    highest_earners = []

    for name, salary in salaries.items():
        if salary > max_salary:
            max_salary = salary
            highest_earners = [name]
        elif salary == max_salary:
            highest_earners.append(name)

    return highest_earners, max_salary


def find_lowest_salary(salaries):
    """Find the lowest earning employee(s) and their salary."""
    if not salaries:
        return None, 0.0

    min_salary = float('inf')
    lowest_earners = []

    for name, salary in salaries.items():
        if salary < min_salary:
            min_salary = salary
            lowest_earners = [name]
        elif salary == min_salary:
            lowest_earners.append(name)

    return lowest_earners, min_salary


def find_above_average_employees(salaries, average_salary):
    """Return a dictionary of employees earning strictly above the average salary."""
    above_average = {}
    for name, salary in salaries.items():
        if salary > average_salary:
            above_average[name] = salary
    return above_average


def display_salary_analysis(salaries):
    """Main function to run all analysis and display formatted results."""
    if not salaries:
        print("No salary data provided.")
        return

    num_employees = len(salaries)
    total_payroll = calculate_total_payroll(salaries)
    avg_salary = calculate_average_salary(total_payroll, num_employees)
    highest_earners, max_salary = find_highest_salary(salaries)
    lowest_earners, min_salary = find_lowest_salary(salaries)
    above_avg_dict = find_above_average_employees(salaries, avg_salary)

    print("\n" + "=" * 45)
    print("        EMPLOYEE SALARY ANALYSIS REPORT      ")
    print("=" * 45)
    print(f"Total Employees          : {num_employees}")
    print(f"Total Payroll            : ${total_payroll:,.2f}")
    print(f"Average Salary           : ${avg_salary:,.2f}")
    print("-" * 45)
    print(f"Highest Salary           : ${max_salary:,.2f} ({', '.join(highest_earners)})")
    print(f"Lowest Salary            : ${min_salary:,.2f} ({', '.join(lowest_earners)})")
    print("-" * 45)
    print("Employees Earning Above Average:")
    if above_avg_dict:
        for name, salary in above_avg_dict.items():
            print(f"  • {name:<20}: ${salary:,.2f}")
    else:
        print("  None")
    print("=" * 45)


def main():
    # Sample employee dataset mapping name -> salary
    employee_salaries = {
        "Alice": 55000.0,
        "Bob": 72000.0,
        "Charlie": 48000.0,
        "Diana": 95000.0,
        "Evan": 62000.0,
        "Fiona": 95000.0  # Tie for highest salary
    }

    display_salary_analysis(employee_salaries)


if __name__ == "__main__":
    main()
```

```python
# 7. Shopping Cart

def add_product(cart):
    """Add a product with its price and quantity to the cart."""
    product_name = input("\nEnter product name: ").strip().title()
    if not product_name:
        print("Product name cannot be empty.")
        return

    try:
        price = float(input(f"Enter price for '{product_name}': $"))
        quantity = int(input(f"Enter quantity for '{product_name}': "))

        if price <= 0 or quantity <= 0:
            print("Price and quantity must be greater than zero.")
            return

        # If product exists, update its quantity
        for item in cart:
            if item["name"].lower() == product_name.lower():
                item["quantity"] += quantity
                print(f"Updated {product_name} quantity to {item['quantity']}.")
                return

        # Add new product
        cart.append({"name": product_name, "price": price, "quantity": quantity})
        print(f"Added {quantity} x {product_name} to cart.")

    except ValueError:
        print("Invalid input! Price must be a number and quantity must be an integer.")


def remove_product(cart):
    """Remove a product from the cart by name."""
    if not cart:
        print("\nCart is empty. Nothing to remove.")
        return

    product_name = input("\nEnter product name to remove: ").strip().lower()

    for item in cart:
        if item["name"].lower() == product_name:
            cart.remove(item)
            print(f"Removed '{item['name']}' from the cart.")
            return

    print(f"Product '{product_name}' not found in cart.")


def view_cart(cart):
    """Display all items in the shopping cart."""
    print("\n" + "=" * 45)
    print("               YOUR SHOPPING CART            ")
    print("=" * 45)
    if not cart:
        print("Your cart is currently empty.")
    else:
        print(f"{'Item':<20} {'Price ($)':<10} {'Qty':<8} {'Total ($)':<10}")
        print("-" * 45)
        for item in cart:
            item_total = item["price"] * item["quantity"]
            print(
                f"{item['name']:<20} {item['price']:<10.2f} {item['quantity']:<8} {item_total:<10.2f}"
            )
    print("=" * 45)


def calculate_bill(cart):
    """Calculate and display the final bill total."""
    view_cart(cart)
    if not cart:
        return

    total_bill = sum(item["price"] * item["quantity"] for item in cart)
    print(f"\nGRAND TOTAL BILL: ${total_bill:.2f}")
    print("=" * 45)


def main():
    """Main application loop."""
    cart = []

    while True:
        print("\n==========================")
        print("    SHOPPING CART MENU    ")
        print("==========================")
        print("1. Add Product")
        print("2. Remove Product")
        print("3. View Cart")
        print("4. Calculate Bill")
        print("5. Exit")

        choice = input("Select an option (1-5): ").strip()

        if choice == "1":
            add_product(cart)
        elif choice == "2":
            remove_product(cart)
        elif choice == "3":
            view_cart(cart)
        elif choice == "4":
            calculate_bill(cart)
        elif choice == "5":
            print("\nThank you for shopping with us! Goodbye.")
            break
        else:
            print("Invalid option! Please enter a number between 1 and 5.")


if __name__ == "__main__":
    main()
```

```python
# 8. Password Strength Checker

import string


def check_password_strength(password):
    """Evaluates a password against security criteria and returns detailed feedback along with a strength level."""
    # Track criteria status using boolean flags
    has_length = len(password) >= 8
    has_uppercase = any(char.isupper() for char in password)
    has_lowercase = any(char.islower() for char in password)
    has_digit = any(char.isdigit() for char in password)
    
    # Define special characters (punctuation set from standard library string module)
    special_characters = set(string.punctuation)
    has_special = any(char in special_characters for char in password)

    # Calculate score based on met criteria
    criteria_met = [has_length, has_uppercase, has_lowercase, has_digit, has_special]
    score = sum(criteria_met)

    # Missing criteria feedback list
    missing_criteria = []
    if not has_length:
        missing_criteria.append("Minimum 8 characters required")
    if not has_uppercase:
        missing_criteria.append("At least one uppercase letter (A-Z)")
    if not has_lowercase:
        missing_criteria.append("At least one lowercase letter (a-z)")
    if not has_digit:
        missing_criteria.append("At least one number (0-9)")
    if not has_special:
        missing_criteria.append("At least one special character (!@#$%^&*, etc.)")

    # Determine qualitative strength rating
    if score == 5:
        strength = "Very Strong"
    elif score == 4:
        strength = "Strong"
    elif score == 3:
        strength = "Moderate"
    elif score == 2:
        strength = "Weak"
    else:
        strength = "Very Weak"

    return {
        "strength": strength,
        "score": f"{score}/5",
        "is_valid": score == 5,
        "missing_criteria": missing_criteria
    }


def display_strength_report(password):
    """Utility function to print formatted password analysis results."""
    result = check_password_strength(password)
    
    print("\n" + "=" * 45)
    print("      PASSWORD STRENGTH ANALYSIS REPORT      ")
    print("=" * 45)
    print(f"Password Tested : {'*' * len(password)}")
    print(f"Strength Level  : {result['strength']}")
    print(f"Criteria Score  : {result['score']}")
    print("-" * 45)

    if result["is_valid"]:
        print("Status          : Valid password! All criteria met.")
    else:
        print("Status          : Insecure password.")
        print("Suggestions to Improve:")
        for rule in result["missing_criteria"]:
            print(f"  • {rule}")
    print("=" * 45)


def main():
    """Main program loop to let users check multiple passwords."""
    while True:
        password = input("\nEnter a password to check (or press Enter to exit): ")
        if not password:
            print("Exiting Password Strength Checker. Goodbye!")
            break

        display_strength_report(password)


if __name__ == "__main__":
    main()
```

```python
# 9. Prime Number Analyzer

def is_prime(n):
    """Check if a number is prime."""
    if n < 2:
        return False
    for i in range(2, int(n**0.5) + 1):
        if n % i == 0:
            return False
    return True


def find_primes(start, end):
    """Find and return all prime numbers in a given range [start, end]."""
    # Swap range boundaries if provided in reverse order
    if start > end:
        start, end = end, start

    primes = []
    for num in range(start, end + 1):
        if is_prime(num):
            primes.append(num)
    return primes


def count_primes(primes_list):
    """Return the total count of prime numbers."""
    return len(primes_list)


def calculate_prime_sum(primes_list):
    """Calculate and return the sum of all prime numbers found."""
    total_sum = 0
    for prime in primes_list:
        total_sum += prime
    return total_sum


def get_largest_prime(primes_list):
    """Return the largest prime number from the list, or None if list is empty."""
    if not primes_list:
        return None
    return primes_list[-1]  # Since range is searched in order, the last element is the largest


def main():
    """Main program loop for analyzing primes in a range."""
    print("=================================")
    print("      PRIME NUMBER ANALYZER      ")
    print("=================================")

    try:
        start_range = int(input("Enter starting number of range: "))
        end_range = int(input("Enter ending number of range: "))

        # Analyze primes within range
        primes = find_primes(start_range, end_range)
        total_count = count_primes(primes)
        total_sum = calculate_prime_sum(primes)
        largest_prime = get_largest_prime(primes)

        # Output Results
        print("\n" + "=" * 35)
        print("          ANALYSIS REPORT        ")
        print("=" * 35)
        print(f"Range Specified : {start_range} to {end_range}")
        print(f"Prime Numbers   : {primes if primes else 'None'}")
        print(f"Total Primes    : {total_count}")
        print(f"Sum of Primes   : {total_sum}")
        print(f"Largest Prime   : {largest_prime if largest_prime is not None else 'None'}")
        print("=" * 35)

    except ValueError:
        print("Invalid input! Please enter valid integers.")


if __name__ == "__main__":
    main()
```

```python
# 10. Expense Tracker

def add_expense(expenses):
    """Add an expense name and amount to the tracking list."""
    name = input("\nEnter expense name: ").strip().title()
    if not name:
        print("Expense name cannot be empty.")
        return

    try:
        amount = float(input(f"Enter amount for '{name}': $"))
        if amount <= 0:
            print("Expense amount must be greater than zero.")
            return

        expenses.append({"name": name, "amount": amount})
        print(f"Successfully added expense: {name} - ${amount:.2f}")

    except ValueError:
        print("Invalid input! Please enter a numerical value for the amount.")


def view_expenses(expenses):
    """Display all logged expenses in a structured format."""
    print("\n" + "=" * 35)
    print("          EXPENSE LIST          ")
    print("=" * 35)
    if not expenses:
        print("No expenses recorded yet.")
    else:
        print(f"{'#':<4} {'Expense Name':<20} {'Amount ($)':<10}")
        print("-" * 35)
        for idx, exp in enumerate(expenses, start=1):
            print(f"{idx:<4} {exp['name']:<20} {exp['amount']:<10.2f}")
    print("=" * 35)


def calculate_total(expenses):
    """Calculate and return the sum of all expenses."""
    total = 0.0
    for exp in expenses:
        total += exp["amount"]

    print(f"\nTotal Expense Amount: ${total:.2f}")
    return total


def find_highest_expense(expenses):
    """Find and display the highest expense record."""
    if not expenses:
        print("\nNo expenses recorded yet.")
        return

    highest = expenses[0]
    for exp in expenses:
        if exp["amount"] > highest["amount"]:
            highest = exp

    print("\n--- Highest Expense ---")
    print(f"Name  : {highest['name']}")
    print(f"Amount: ${highest['amount']:.2f}")


def main():
    """Main application loop."""
    expenses = []

    while True:
        print("\n==========================")
        print("     EXPENSE TRACKER      ")
        print("==========================")
        print("1. Add Expense")
        print("2. View Expenses")
        print("3. Calculate Total")
        print("4. Find Highest Expense")
        print("5. Exit")

        choice = input("Select an option (1-5): ").strip()

        if choice == "1":
            add_expense(expenses)
        elif choice == "2":
            view_expenses(expenses)
        elif choice == "3":
            calculate_total(expenses)
        elif choice == "4":
            find_highest_expense(expenses)
        elif choice == "5":
            print("\nExiting Expense Tracker. Goodbye!")
            break
        else:
            print("Invalid option! Please select a number between 1 and 5.")


if __name__ == "__main__":
    main()
```

```python
# 11. Mini Authentication System

# Predefined credentials stored in a dictionary
CREDENTIALS = {
    "admin": "Admin@123",
    "john_doe": "Password456",
    "alice": "SecurePass789"
}

MAX_ATTEMPTS = 3


def authenticate_user(username, password):
    """Verify if username and password match predefined credentials."""
    if username in CREDENTIALS and CREDENTIALS[username] == password:
        return True
    return False


def login():
    """Handle login loop, tracks attempts, and manages failed login/retry logic."""
    attempts = 0

    print("\n" + "=" * 35)
    print("       USER AUTHENTICATION       ")
    print("=" * 35)

    while attempts < MAX_ATTEMPTS:
        username = input("Enter username: ").strip()
        password = input("Enter password: ").strip()

        if authenticate_user(username, password):
            print(f"\nLogin Successful! Welcome, {username}.")
            return username  # Return logged-in username
        else:
            attempts += 1
            remaining = MAX_ATTEMPTS - attempts
            print("\n[Failed Login] Invalid username or password.")

            if remaining > 0:
                print(f"Attempts remaining: {remaining}. Please try again.\n")
            else:
                print("\n[Account Locked] Maximum login attempts exceeded!")
                return None


def user_dashboard(username):
    """Display authenticated session options and handle logout."""
    while True:
        print("\n" + "=" * 35)
        print(f"    LOGGED IN SESSION: {username.upper()}    ")
        print("=" * 35)
        print("1. View Profile")
        print("2. Dashboard Settings")
        print("3. Logout")

        choice = input("Select an option (1-3): ").strip()

        if choice == "1":
            print(f"\nUser Profile: {username}")
        elif choice == "2":
            print("\nSettings: Account is active.")
        elif choice == "3":
            print(f"\nLogging out user '{username}'...")
            print("Successfully logged out!")
            break
        else:
            print("Invalid selection! Please choose 1, 2, or 3.")


def main():
    """Main application loop allowing login attempts and application exit."""
    while True:
        print("\n==========================")
        print("   AUTHENTICATION SYSTEM  ")
        print("==========================")
        print("1. Login")
        print("2. Exit Application")

        choice = input("Select an option (1-2): ").strip()

        if choice == "1":
            logged_in_user = login()
            if logged_in_user:
                # Redirect to authenticated dashboard until logged out
                user_dashboard(logged_in_user)
        elif choice == "2":
            print("\nExiting application. Goodbye!")
            break
        else:
            print("Invalid option! Please enter 1 or 2.")


if __name__ == "__main__":
    main()
```

```python
import string


# ==========================================
# UTILITY 1: CALCULATOR
# ==========================================
def calculator():
    """Performs basic arithmetic operations (+, -, *, /)."""
    print("\n--- [1] Calculator ---")
    try:
        num1 = float(input("Enter first number: "))
        operator = input("Enter operator (+, -, *, /): ").strip()
        num2 = float(input("Enter second number: "))

        if operator == '+':
            result = num1 + num2
        elif operator == '-':
            result = num1 - num2
        elif operator == '*':
            result = num1 * num2
        elif operator == '/':
            if num2 == 0:
                print("Error: Division by zero is not allowed.")
                return
            result = num1 / num2
        else:
            print("Invalid operator!")
            return

        print(f"Result: {num1} {operator} {num2} = {result:.2f}")
    except ValueError:
        print("Invalid input! Please enter valid numeric values.")


# ==========================================
# UTILITY 2: PALINDROME CHECKER
# ==========================================
def palindrome_checker():
    """Checks whether a string or number is a palindrome."""
    print("\n--- [2] Palindrome Checker ---")
    text = input("Enter a word, sentence, or number: ").strip().lower()
    
    # Filter out spaces and non-alphanumeric characters for accurate word checks
    cleaned_text = "".join(char for char in text if char.isalnum())

    if not cleaned_text:
        print("Empty input. Nothing to check.")
        return

    if cleaned_text == cleaned_text[::-1]:
        print(f"Result: '{text}' IS a palindrome!")
    else:
        print(f"Result: '{text}' is NOT a palindrome.")


# ==========================================
# UTILITY 3: PRIME CHECKER
# ==========================================
def prime_checker():
    """Checks whether a number is prime or composite."""
    print("\n--- [3] Prime Checker ---")
    try:
        num = int(input("Enter an integer: "))
        if num < 2:
            print(f"Result: {num} is NOT a prime number.")
            return

        is_prime = True
        for i in range(2, int(num ** 0.5) + 1):
            if num % i == 0:
                is_prime = False
                break

        if is_prime:
            print(f"Result: {num} IS a prime number!")
        else:
            print(f"Result: {num} is NOT a prime number.")
    except ValueError:
        print("Invalid input! Please enter a valid integer.")


# ==========================================
# UTILITY 4: FACTORIAL CALCULATOR
# ==========================================
def factorial_calculator():
    """Calculates the factorial of a non-negative integer using a loop."""
    print("\n--- [4] Factorial Calculator ---")
    try:
        num = int(input("Enter a non-negative integer: "))
        if num < 0:
            print("Factorial is not defined for negative numbers.")
            return

        factorial = 1
        for i in range(1, num + 1):
            factorial *= i

        print(f"Result: {num}! = {factorial}")
    except ValueError:
        print("Invalid input! Please enter a valid integer.")


# ==========================================
# UTILITY 5: MULTIPLICATION TABLE
# ==========================================
def multiplication_table():
    """Generates a multiplication table for a number up to a specified range."""
    print("\n--- [5] Multiplication Table Generator ---")
    try:
        num = int(input("Enter table number: "))
        limit = int(input("Enter range limit (e.g., 10 or 12): "))

        print(f"\nMultiplication Table for {num}:")
        print("-" * 25)
        for i in range(1, limit + 1):
            print(f"{num} x {i:<2} = {num * i}")
        print("-" * 25)
    except ValueError:
        print("Invalid input! Please enter valid integers.")


# ==========================================
# UTILITY 6: PASSWORD CHECKER
# ==========================================
def password_checker():
    """Evaluates password strength based on length and character types."""
    print("\n--- [6] Password Checker ---")
    password = input("Enter password to check: ")

    has_len = len(password) >= 8
    has_upper = any(c.isupper() for c in password)
    has_lower = any(c.islower() for c in password)
    has_digit = any(c.isdigit() for c in password)
    has_special = any(c in string.punctuation for c in password)

    score = sum([has_len, has_upper, has_lower, has_digit, has_special])

    print(f"\nPassword Strength Score: {score}/5")
    if score == 5:
        print("Rating: Strong / Valid Password")
    elif score >= 3:
        print("Rating: Moderate Password")
    else:
        print("Rating: Weak Password")


# ==========================================
# MAIN APPLICATION MENU
# ==========================================
def main():
    """Main menu loop providing access to all utilities."""
    while True:
        print("\n===========================================")
        print("    SUPER30 PYTHON UTILITY APPLICATION     ")
        print("===========================================")
        print("1. Calculator")
        print("2. Palindrome Checker")
        print("3. Prime Checker")
        print("4. Factorial Calculator")
        print("5. Multiplication Table")
        print("6. Password Checker")
        print("7. Exit")

        choice = input("Select a utility (1-7): ").strip()

        if choice == '1':
            calculator()
        elif choice == '2':
            palindrome_checker()
        elif choice == '3':
            prime_checker()
        elif choice == '4':
            factorial_calculator()
        elif choice == '5':
            multiplication_table()
        elif choice == '6':
            password_checker()
        elif choice == '7':
            print("\nThank you for using Super30 Utility Application. Goodbye!")
            break
        else:
            print("Invalid choice! Please select an option between 1 and 7.")


if __name__ == "__main__":
    main()
```
