```python
#  print numbers from 1 to 100 using a for loop.
# Print a header message to describe the output
print(" Displaying numbers from 1-100:")

# Loop through numbers from 1 up to 100.
# range(1, 101) starts at 1 and stops before 101 (since the end boundary is exclusive).
for i in range(1, 101):
    # Output the current value of i for each iteration
    print(i)
```

Logic:

range(1, 101): Python's range(start, stop) function generates numbers starting from start up to, but not including, stop. Using 101 ensures the loop includes 100.

for i in ...: Assigns each generated integer to the variable i sequentially on each iteration.

print(i): Displays the current integer i to the console.

```python
#  print numbers from 1 to 100 using a for loop.
# Set the upper limit for the range
n = 100

# range(1, n + 1) generates a sequence starting from 1 up to n (100) inclusive,
# because range() excludes the end boundary value (101).
for i in range(1, n + 1):
    # Print the current number in the loop iteration
    print(i)
```

Logic:

n = 100: Defines the target maximum value.

range(1, n + 1): range(start, stop) stops before reaching stop. Passing 101 (100 + 1) ensures 100 is included in the iteration.

for i in ...: Assigns each value in the range to variable i sequentially during each loop iteration.

```python
#  print all even numbers from 1 to 100 using a for loop.

# Print a header message describing the output
print("Even numbers from 1 to 100:")

# range(start, stop, step) begins at 2, increments by 2 each iteration, 
# and stops before 101 to include 100.
for i in range(2, 101, 2):
    # Output the current even number
    print(i)
```

Logic:

range(2, 101, 2):

start = 2: Starts directly at the first even number.

stop = 101: Excludes 101, ensuring 100 is included in the output.

step = 2: Skips odd numbers completely by adding 2 on every iteration (2, 4, 6, ..., 100), making this more efficient than using a modulo filter (i % 2 == 0).

print(i): Outputs each even number sequentially.

```python
#  print all even numbers from 1 to 100 using a for loop.

# Print a header message to describe the output
print("Even numbers from 1 to 100:")

# Loop through all numbers starting from 1 up to 100 (101 is exclusive)
for i in range(1, 101):
    # Check if the current number is even using the modulo operator (%).
    # i % 2 evaluates to 0 if the remainder after dividing by 2 is zero.
    if i % 2 == 0:
        # Print the number only if the condition evaluates to True
        print(i)
```

i % 2 == 0: The modulo operator (%) calculates the remainder of i divided by 2. If the remainder is 0, the number is even.

if Condition: Filters the loop execution so print(i) runs only when the modulo check succeeds.

```python
#  print all odd numbers from 1 to 100 using a for loop.

# Print a header message describing the output
print("Odd numbers from 1 to 100:")

# Loop through all numbers starting from 1 up to 100 (101 is exclusive)
for i in range(1, 101):
    # Check if the current number is odd using the modulo operator (%).
    # i % 2 evaluates to non-zero (1) if the number is odd.
    if i % 2 != 0:
        # Output the current odd number
        print(i)
```

```python
#  print all even and odd numbers from 1 to 100 in two separate lists.

# Define a sequence of numbers from 1 to 100 (101 is exclusive)
l = range(1, 101)

# Initialize empty lists to store even and odd numbers
l_even = []
l_odd = []

# Loop through each number in the range
for i in l:
    # Check if the current number is even using the modulo operator (%)
    if i % 2 == 0:
        # Append even numbers to the l_even list
        l_even.append(i)
    else:
        # Append odd numbers to the l_odd list
        l_odd.append(i) 

# Print both populated lists
print("Even numbers:", l_even)
print("Odd numbers:", l_odd) 
```

Logic:

l_even = [] / l_odd = []: Prepares empty container lists before iteration to store filtered values dynamically.

if i % 2 == 0: Determines parity. If the remainder of i divided by 2 is 0, it routes to the even list.

else: Automatically catches any number that fails the even check (which must be odd) and routes it to the odd list.

.append(i): Adds the current integer to the tail of the corresponding list during each iteration.

```python
#  take an integer n and print its multiplication table from 1 to 20.

# Prompt the user to input an integer and convert the string input to an integer
n = int(input("Enter an integer: "))

# Print a header for clarity
print(f"Multiplication Table for {n}:")

# Loop through numbers from 1 to 20 inclusive
# range(1, 21) generates numbers 1, 2, ..., 20 (stopping before 21)
for i in range(1, 21):
    # Print the equation in the format: n x i = product
    print(n, "x", i, "=", n * i)
```

```python
#  calculate the sum of numbers from 1 to n using a loop.

# Initialize a accumulator variable to keep track of the running total
total_sum = 0

# Prompt the user to enter the target upper bound and convert it to an integer
n = int(input("Enter range you want to sum: "))

# Loop through every integer from 1 up to n inclusive
# range(1, n + 1) ensures n is included since the end boundary is exclusive
for i in range(1, n + 1):
    # Add the current integer i to the running total
    total_sum = total_sum + i  # Alternatively: total_sum += i

# Output the final accumulated sum
print("The sum of numbers from 1 to", n, " is:", total_sum)
```

Logic:

total_sum = 0: Initializes the running sum accumulator variable before entering the loop. (Note: Avoid using sum as a variable name in Python to prevent overwriting the built-in sum() function).

range(1, n + 1): Generates sequence values starting at 1 and terminating at n (since n + 1 is non-inclusive).

total_sum += i: Shorthand for total_sum = total_sum + i, incrementally adding the current value of i to the running total on each iteration.

```python
#  calculate the factorial of a number without using built-in factorial functions.

# Prompt the user to enter a non-negative integer and convert it to an int
number = int(input("enter the number to find out factorial:"))

# Initialize an accumulator variable to 1 (since 0! = 1 and 1 is the multiplicative identity)
fact = 1

# Loop through all integers from 1 up to 'number' inclusive
# range(1, number + 1) generates numbers 1, 2, ..., number
for i in range(1, number + 1):
    # Multiply the running factorial product by the current integer i
    fact = fact * i  # Alternatively: fact *= i

# Output the calculated factorial result
print("The factorial of", number, "is:", fact)
```

Logic:

fact = 1: Factorial accumulation must start at 1. Starting at 0 would cause every subsequent multiplication to result in 0.range(1, number + 1): Generates every integer starting from 1 up to number. Adding 1 ensures the upper bound is included.fact = fact * i: Sequentially multiplies the accumulated result by each integer in the range (e.g., $1 \times 1 \times 2 \times 3 \times \dots \times n$).

```python
# Count how many vowels exist in a user-provided string.

# Prompt the user to enter a string
string = input("Enter the string to check vowels: ")

# Initialize a counter variable to keep track of the total vowels found
vowel = 0

# Loop through each character in the input string
for i in string:
    # Check if the current character is a vowel (both lowercase and uppercase)
    if i in 'aeiouAEIOU':
        # Print the identified vowel character
        print(i, "is a vowel")
        # Increment the vowel counter by 1
        vowel = vowel + 1  # Alternatively: vowel += 1

# Display the total count of vowels found
print("Total number of vowels in the string is:", vowel)
  
```

vowel = vowel + 1: Increments the total count each time the conditional check evaluates to True.

```python
#  reverse a string using a for loop without using [::-1] or reversed().

# Prompt the user to enter a string to reverse
string1 = input("Enter the string: ")

# Calculate the total length of the input string
length = len(string1)

# Initialize an empty string variable to accumulate the reversed characters
rev_string = ""

# Display the length of the original string
print("The length of the string is:", length)

# Loop through the indices from 0 up to length - 1
for i in range(length):
    # Print the character being accessed from the back of the string
    print("The character is:", string1[length - 1 - i])
    
    # Print the current loop index
    print("value of i:", i)
    
    # Print the calculated index corresponding to the reverse position
    print("value of lenghth-1-i:", length - 1 - i)
    
    # Concatenate the character at the calculated position to the reversed string
    rev_string = rev_string + string1[length - 1 - i]

# Output the final reversed string result
print("The reverse of the string is:", rev_string)
```

Logic:

length - 1 - i: Reverses the indexing manually without using [::-1] or reversed(). When i = 0, it accesses the last index (length - 1). As i increases, it steps backward toward index 0.

rev_string = rev_string + ...: Appends each character from back-to-front to build the reversed string step-by-step.

range(length): Ensures the loop runs exactly once for every character position in the string.

```python
# Find the largest number from a list without using max().

# Given list of numbers
numbers = [3, 7, 2, 9, 1, 5]

# Assume the first element is the largest initially
largest = numbers[0]

# Iterate through each element in the list
for i in numbers:
    # Trace values during iteration
    print("i value is:", i)
    print("The current largest number is:", largest)
    
    # Compare current element with the tracked maximum
    if i > largest:
        # Update largest if a bigger number is found
        largest = i

print("x" * 33)
# Output the final largest number found
print("The largest number is:", largest)
```

# Logic:

largest = numbers[0]: Initializes largest with the list's first element as a benchmark, ensuring it works even with negative numbers.

for i in numbers: Loops sequentially through every item in the list.

if i > largest: Evaluates whether the current number i is greater than the recorded maximum.

largest = i: Replaces the value of largest whenever a larger value is encountered during iteration.
