```python
# Create functions for addition subtraction multiplication division  

def add(a, b):
    """Return the addition of two numbers."""
    # Add two numbers together and return the result
    return a + b

def subtract(a, b):
    """Return the subtraction of two numbers."""
    # Subtract b from a and return the result
    return a - b

def multiply(a, b):
    """Return the product of two numbers."""
    # Multiply two numbers together and return the result
    return a * b

def divide(a, b):
    """Return the division of two numbers, checking for zero division."""
    # Check for division by zero before dividing
    if b == 0:
        return "Error: Division by zero is not allowed."
    # Divide a by b and return the result
    return a / b    

# Display menu options to the user
print("Select operation:")
print("1. Add")
print("2. Subtract")
print("3. Multiply")
print("4. Divide")  

# Prompt user for their choice
print("Enter choice (1/2/3/4):")
choice = input()

# Check if the user's choice is valid before prompting for numbers
if choice in ('1', '2', '3', '4'):
    # Prompt user for the first number and convert to float
    print("Enter first number:")
    num1 = float(input())

    # Prompt user for the second number and convert to float
    print("Enter second number:")
    num2 = float(input())   

    # Execute operation based on user choice
    print("Result:")
    if choice == '1':
        # Call add function and print the formatted result
        print(num1, "+", num2, "=", add(num1, num2))
    elif choice == '2':
        # Call subtract function and print the formatted result
        print(num1, "-", num2, "=", subtract(num1, num2))
    elif choice == '3':         
        # Call multiply function and print the formatted result
        print(num1, "*", num2, "=", multiply(num1, num2))
    elif choice == '4':
        # Call divide function and print the formatted result
        print(num1, "/", num2, "=", divide(num1, num2))
else:
    # Handle invalid menu selections immediately
    print("Invalid input")
```

```python
# Create a function that returns the largest of three numbers without using max().

def find_largest(a, b, c):
    """Return the largest of three numbers using conditional statements."""
    # Compare 'a' against 'b' and 'c'
    if a >= b and a >= c:
        return a
    # Compare 'b' against 'a' and 'c'
    elif b >= a and b >= c:
        return b
    # If neither 'a' nor 'b' is the largest, 'c' must be the largest
    else:
        return c

# Display instruction message to user
print("Enter three numbers to find the largest:")

# Prompt user for the first number and convert input to float
num1 = float(input("Enter first number: "))

# Prompt user for the second number and convert input to float
num2 = float(input("Enter second number: "))

# Prompt user for the third number and convert input to float
num3 = float(input("Enter third number: "))

# Call find_largest function with user inputs
largest = find_largest(num1, num2, num3)

# Print the determined largest number
print("The largest number is:", largest)
```

```python
# Create a function that calculates factorial.

def calculate_factorial(n):
    """Calculate the factorial of a given non-negative integer."""
    # Check if the number is negative since factorial is undefined for negative numbers
    if n < 0:
        return "Error: Factorial is not defined for negative numbers."
    # The factorial of 0 and 1 is always 1
    elif n == 0 or n == 1:
        return 1
    else:
        # Initialize the factorial result variable to 1
        factorial = 1
        # Loop through numbers from 2 up to n to compute the factorial
        for i in range(2, n + 1):
            factorial *= i
        # Return the final calculated factorial value
        return factorial

# Prompt the user for input
print("Enter a number to calculate its factorial:")

# Read the user input and convert it to an integer
number = int(input())

# Pass the integer to calculate_factorial and store the output
result = calculate_factorial(number)

# Output the calculated factorial or error message to the console
print(result)
```

```python
# Create a function that calculates factorial - another logic -Recursive

def factorial(n):
    """Calculate the factorial of a given non-negative integer using recursion."""
    # Factorial is undefined for negative numbers
    if n < 0:
        return "Error: Factorial is not defined for negative numbers."
    # Base case: factorial of 0 or 1 is 1
    elif n == 0 or n == 1:
        fact = 1
        return fact
    # Recursive case: n * factorial of (n - 1)
    else:
        fact = n * factorial(n - 1)
        return fact

# Prompt the user for input
print("Enter a number to calculate its factorial:")

# Read user input and convert it to an integer
number = int(input())

# Call the factorial function with user input
result = factorial(number)

# Print the resulting factorial or error message
print(result)
```

```python
# Create a function that checks whether a number is prime. 

def is_prime(n):
    """Check whether a given integer is a prime number."""
    # Numbers less than 2 are not prime numbers
    if n < 2:
        return False
    
    # Flag to track whether the number is prime
    is_prime = True
    
    # Check for divisibility from 2 up to n - 1
    for i in range(2, n):
        if n % i == 0:
            is_prime = False
            break
            
    # Return the boolean result
    return is_prime

# Prompt the user to enter a number
print("Enter a number to check if it's prime:")

# Read user input and convert it to an integer
number = int(input())

# Call the is_prime function and print the appropriate message
if is_prime(number):
    print(f"{number} is a prime number.")
else:
    print(f"{number} is not a prime number.")
```

```python
# Create def calculate_discount(price, discount=10) where the default discount is 10%.
def calculate_discount(price, discount=10):
    """Calculate the discounted price of an item.
    
    Parameters:
    price (float): The original price of the item.
    discount (float): The discount percentage (default is 10%).
    
    Returns:
    float: The discounted price.
    """
    # Calculate final price after subtracting the discount amount
    discounted_price = price - (price * discount / 100)
    return discounted_price

# Example usage:
# Prompt user to input the item's original price and convert it to float
original_price = float(input("Enter the original price of the item: "))

# Call calculate_discount with default 10% discount parameter
discounted_price = calculate_discount(original_price)

# Output the formatted discounted price to two decimal places
print(f"The discounted price is: ${discounted_price:.2f}")
```

```python
# Create a function that accepts a list and returns its sum without using sum(). 

def calculate_sum(numbers):
    """Calculate the sum of a list of numbers without using the built-in sum() function."""
    # Initialize a variable to accumulate the total sum
    total = 0
    # Iterate through each number in the list and add it to total
    for num in numbers:
        total += num
    # Return the computed sum
    return total

# Prompt user to specify how many numbers they wish to enter
print("Enter the count of numbers you want to sum:")

# Read the count and convert it to an integer
count = int(input())

# Initialize an empty list to store user inputs
numbers = []

# Loop 'count' times to collect each number from the user
for i in range(count):
    num = float(input(f"Enter number {i + 1}: "))
    numbers.append(num)

# Call calculate_sum with the list of numbers
result = calculate_sum(numbers)

# Output list metadata and final calculated sum
print("the length of the list is:", len(numbers))
print("the entered list is:", numbers)
print("The sum of the list is:", result)
```

```python
# Create a function that accepts a string and returns the number of vowels.

def count_vowels(s):
    """Count and return the total number of vowels in a given string."""
    # Define a string containing all uppercase and lowercase vowels
    vowels = "aeiouAEIOU"
    
    # Initialize a counter variable to track the vowel count
    count = 0
    
    # Iterate through each character in the input string
    for char in s:
        # Check if the character is a vowel
        if char in vowels:
            count += 1
            
    # Return the final count of vowels
    return count

# Example usage:
# Prompt user to input a string
input_string = input("Enter a string: ")

# Pass the input string to the function and store the returned count
vowel_count = count_vowels(input_string)

# print(f"The number of vowels in the string '{input_string}' is: {vowel_count}")

# Print the resulting vowel count along with the original input string
print("The number of vowels in the string" ,input_string ,"is:", vowel_count)
```

```python
# Create a function that accepts name age course and returns a formatted student profile. Use both positional and keyword arguments while calling it.

def student_profile(name, age, course):
    """Return a formatted string representing a student's profile."""
    # Format and return student details using an f-string
    return f"Student Name: {name}\nAge: {age}\nCourse: {course}"

# Example usage with positional arguments:
# Call function passing arguments by position
print(student_profile("Alice", 20, "Computer Science"))

# Example usage with keyword arguments:
# Call function passing arguments explicitly by keyword name
print(student_profile(name="Bob", age=22, course="Mathematics"))
```

```python
# Demonstrate the difference between a function that uses print() a function that uses return 
# Explain through code why return is useful. 

def print_profile(name, age, course):
    """Print the student profile details directly to the console."""
    # Output name, age, and course directly to standard output
    print(f"Student Name: {name}")
    print(f"Age: {age}")
    print(f"Course: {course}")


def return_profile(name, age, course):
    """Return a formatted student profile string."""
    # Format and return the student profile string
    return f"Student Name: {name}\nAge: {age}\nCourse: {course}"


# Example usage:
# Call print_profile to output details directly
print_profile("Alice", 20, "Computer Science")

# Print a blank line for separation
print()

# Print the string returned by return_profile
print(return_profile("Alice", 20, "Computer Science"))

# Theoretical explanation of print() vs return
explanation = """  
In the first function, print_profile(), the information is printed directly to the console.
In the second function, return_profile(), the information is returned as a string, which can be 
stored in a variable, printed, or used in further computations. 
This makes return more versatile than print

print() displays output to the console for a human to see, but returns None to the program. 
return hands a value back to the caller so it can be saved in a variable, passed to other functions, 
or used in further calculations.
"""


def add_with_print(a, b):
    """Print the sum of two numbers directly to the console (returns None)."""
    # Print the calculated sum directly
    print(a + b)


def add_with_return(a, b):
    """Return the sum of two numbers so it can be re-used elsewhere."""
    # Return the calculated sum
    return a + b


# --- 1. Basic Function Call Output ---
print("--- Using print() ---")
# Call add_with_print which outputs 10 to the console directly
add_with_print(5, 5)

print("\n--- Using return ---")
# Store the return value of add_with_return in the variable 'result'
result = add_with_return(5, 5)

# Print the stored result variable
print(result)


# Practical explanation of return's utility
return_usefulness = """
Why return is Useful:
-----------------------
return makes functions modular and reusable. When a function returns a value, that result can be 
chained into further logic, which is impossible with print().
"""


# --- Scenario: Calculate a 10% tax on the result ---

# ATTEMPT A: Using print() - FAILS
def calculate_tax_fail():
    """Demonstrate how using a function with print() fails in further calculations."""
    # Prints 150, but 'val' receives None because add_with_print returns None
    val = add_with_print(100, 50)
    # The line below fails with:
    # TypeError: unsupported operand type(s) for *: 'NoneType' and 'float'
    # tax = val * 0.10


# ATTEMPT B: Using return - WORKS
def calculate_tax_success():
    """Calculate and return final total including a 10% tax using a return value."""
    # 'val' receives the returned integer 150
    val = add_with_return(100, 50)
    
    # Calculate 10% tax on the returned value
    tax = val * 0.10
    
    # Calculate final total sum
    final_total = val + tax
    
    # Return the final calculated total
    return final_total


# Execute and display the result of the successful tax calculation
print("Final Total with Tax:", calculate_tax_success())


# Summary comparison notes
summary = """
Key Differences:
--------------------
print(): Used for debugging, user interfaces, or logs. Its output cannot be stored or processed 
further by Python code.

return: Used to send data back to the rest of your program. It enables mathematical operations, 
string formatting, and chaining multiple functions together.
"""
```
