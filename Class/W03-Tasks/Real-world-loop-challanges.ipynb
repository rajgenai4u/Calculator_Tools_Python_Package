```python
# calculate total transaction value without using sum().

# Define the list of transaction values
transactions = [1200, 450, 800, 1500, 2300, 700, 100]

# Initialize an accumulator variable to store the running sum
total = 0

# Iterate through each transaction amount in the list
for i in transactions:
    # Add the current transaction value to the running total
    total += i

# Display the total calculated transaction value
print(total)
```

```python
# Find highest transaction value without using max().

# Initialize big with the first element of the transactions list as the baseline comparison
big = transactions[0]

# Iterate through each transaction amount in the list
for i in transactions:
    # Compare the current transaction with the highest value found so far
    if i > big: 
        # Update big if the current value is greater
        big = i

# Display the highest transaction value found
print(big)
```

```python
# Find lowest transaction value without using min().

# Initialize lowest with the first element of the transactions list as the baseline comparison
lowest = transactions[0]

# Iterate through each transaction amount in the list
for i in transactions:
    # Compare the current transaction with the lowest value found so far
    if i < lowest:
        # Update lowest if the current value is smaller
        lowest = i

# Display the lowest transaction value found
print(lowest)
```

```python
# find the average temperature without using sum() or len().

# Define the list of temperatures
temperatures = [32, 35, 28, 40, 38, 31, 42]

# Initialize trackers for element count (length) and sum (total_temp)
length = 0
total_temp = 0

# Iterate through each temperature in the list
for i in temperatures:
    # Accumulate the total temperature sum
    total_temp += i
    
    # Manually count the number of items
    length += 1

# Calculate the average by dividing total sum by total count
average_temp = total_temp / length

# Display the calculated average temperature
print(average_temp)
```

```python
# find the average temperature using sum() or len().

# Calculate average by dividing the sum of elements by the number of elements
avg_temp = sum(temperatures) / len(temperatures)

# Display the calculated average temperature
print(avg_temp)
```

```python
# Count count how many students scored 90+
# students marks list
marks = [78, 92, 45, 67, 88, 53, 99]

# Initialize counter for marks 90 and above
count_90_plus = 0

# Iterate through each student's mark in the list
for i in marks:
    # Check if the mark is greater than or equal to 90
    if i >= 90:
        # Increment the 90+ counter by 1
        count_90_plus += 1

# Display the total count of students scoring 90+
print(count_90_plus)
```

```python
# Count count how many students scored 90+ using list comprehension
count_90_plus_lc = len([i for i in marks if i >= 90])

# Display the total count of students scoring 90+
print(count_90_plus_lc)
```

```python
# Use filter() and a lambda function to keep only marks that are 90 or above,
# convert the filtered result to a list, and calculate its length to get the total count
count_90_plus_filter = len(list(filter(lambda x: x >= 90, marks)))

# Print the final count of students who scored 90+
print(count_90_plus_filter)
```

```python
# Create a simple login system with a maximum of 3 password attempts.

# Set the correct predefined password
password = "secret123"

# Define the maximum allowed login attempts
max_attempts = 3

# Initialize the attempt counter variable
attempts = 0

# Loop as long as the user has remaining attempts
while attempts < max_attempts:
    # Prompt the user to enter their password
    user_input = input("Enter password: ")
    
    # Check if the entered password matches the correct password
    if user_input == password:
        # Inform the user of a successful login
        print("Login successful!")
        # Exit the loop immediately upon successful login
        break
    else:
        # Increment the attempt counter on incorrect password entry
        attempts += 1
        # Display an error message showing the remaining attempts left
        print("Invalid password", max_attempts - attempts, "attempts remaining.")
# The else block executes only if the while loop completes without hitting a 'break' statement
else:
    # Inform the user that they have been locked out due to exceeding maximum attempts
    print("Too many failed attempts. Access denied.")
```

```python
# Function to count character frequencies in a given string without using Counter
def count_char_frequency(input_string):
    # Initialize an empty dictionary to store each character as a key and its count as the value
    char_counts = {}

    # Iterate through each character in the input string
    for char in input_string:
        # If the character is already a key in the dictionary, increment its count by 1
        if char in char_counts:
            char_counts[char] += 1
        # If the character is not yet in the dictionary, add it with an initial count of 1
        else:
            char_counts[char] = 1

    # Return the dictionary containing all character counts
    return char_counts


# Define the input string for testing
text = "banana"

# Call the function and store the returned dictionary of character counts
frequencies = count_char_frequency(text)

# Iterate through key-value pairs in the dictionary to display each character and its count
for char, count in frequencies.items():
    # Print each character and its corresponding frequency formatted as "char -> count"
    print(f"{char} -> {count}")
```

```python
# Function to find the second-largest number in a list without using sort()
def find_second_largest(numbers):
    # Return None if the list has fewer than 2 elements since a second largest cannot exist
    if len(numbers) < 2:
        return None

    # Initialize largest and second_largest with initial values from the list
    largest = numbers[0]
    second_largest = numbers[1]

    # Iterate through every number in the input list
    for num in numbers:
        # If the current number is greater than the current largest:
        # update second_largest to the previous largest, and update largest to the current number
        if num > largest:
            second_largest = largest
            largest = num
        # If the current number is greater than second_largest but not equal to largest:
        # update second_largest to the current number
        elif num > second_largest and num != largest:
            second_largest = num

    # Return second_largest if a distinct second-largest value exists; otherwise, return None
    return second_largest if second_largest != largest else None


# Example usage with a list of numbers
numbers = [10, 45, 2, 99, 175, 45, 87, 150]

# Call the function and store the result
result = find_second_largest(numbers)

# Print the second-largest number found in the list
print(f"Second largest number: {result}")
```

```python
# Function to find the second-largest number in a list without using sort()
def find_second_largest(numbers):
    # Check if the list contains fewer than 2 elements
    if len(numbers) < 2:
        return None

    # Initialize tracking variables to None to handle both positive and negative numbers correctly
    largest = None
    second_largest = None

    # Loop through each number in the list
    for num in numbers:
        # If largest is not set or current number is greater than the current largest
        if largest is None or num > largest:
            second_largest = largest  # Shift previous largest down to second_largest
            largest = num              # Set current number as the new largest
        # Otherwise, if num is not equal to largest and is strictly greater than second_largest
        elif num != largest and (second_largest is None or num > second_largest):
            second_largest = num      # Set current number as the new second_largest

    # Return the second largest value found (or None if no distinct second largest exists)
    return second_largest


# Example usage with a list of numbers containing duplicates and unsorted values
numbers = [10, 45, 2, 99, 105, 45, 87, 100]

# Commented out test case for edge case where all numbers are identical:
# numbers = [10, 10, 10, 10, 10, 10, 10, 10]

# Call the function with the input list
result = find_second_largest(numbers)

# Print the result
print(f"Second largest number: {result}")
```

```python
# Function to check if a string is a palindrome using two pointers and a while loop
def is_palindrome(text):
    # Convert string to lowercase to ensure case-insensitive comparison
    cleaned_text = text.lower()

    # Initialize two pointers: left starting at the beginning, right at the end
    left = 0
    right = len(cleaned_text) - 1

    # Compare characters from both ends moving inward toward the center
    while left < right:
        # If characters at current pointers do not match, it's not a palindrome
        if cleaned_text[left] != cleaned_text[right]:
            return False
        # Move the left pointer one step to the right
        left += 1
        # Move the right pointer one step to the left
        right -= 1

    # Return True if all corresponding characters matched
    return True


# List of sample words to test the function
test_words = ["radar", "Madam", "banana", "racecar", "python"]

# Loop through each word in the test list
for word in test_words:
    # Check if the word is a palindrome using the function
    if is_palindrome(word):
        # Print confirmation if it is a palindrome
        print(f"'{word}' is a palindrome.")
    else:
        # Print confirmation if it is not a palindrome
        print(f"'{word}' is NOT a palindrome.")
```

```python
# Function to display the user's current account balance formatted to 2 decimal places
def check_balance(balance):
    print(f"\nYour current balance is: ${balance:.2f}")


# Function to handle money deposit operations
def deposit_money(balance):
    # Prompt user for deposit amount and convert to float
    amount = float(input("Enter amount to deposit: $"))
    # Validate that deposit amount is greater than zero
    if amount <= 0:
        print("Invalid deposit amount. Please enter a positive value.")
        return balance

    # Add the deposited amount to the total balance
    balance += amount
    print(f"Successfully deposited ${amount:.2f}.")
    print(f"Updated balance: ${balance:.2f}")
    return balance


# Function to handle money withdrawal operations
def withdraw_money(balance):
    # Prompt user for withdrawal amount and convert to float
    amount = float(input("Enter amount to withdraw: $"))
    # Validate that the withdrawal amount is positive
    if amount <= 0:
        print("Invalid withdrawal amount. Please enter a positive value.")
    # Check if the requested amount exceeds the available account balance
    elif amount > balance:
        print(f"Insufficient funds! Your current balance is ${balance:.2f}.")
    # Deduct the amount from balance if funds are sufficient
    else:
        balance -= amount
        print(f"Successfully withdrew ${amount:.2f}.")
        print(f"Updated balance: ${balance:.2f}")

    return balance


# Main function to run the ATM user interface loop
def atm_simulation():
    # Set starting balance for the simulation
    balance = 1000.0  # Initial balance

    # Loop continuously to allow repeated transactions until user exits
    while True:
        # Display the main menu choices
        print("\n=== ATM SIMULATION ===")
        print("1. Check Balance")
        print("2. Deposit Money")
        print("3. Withdraw Money")
        print("4. Exit")

        # Get user's menu choice and trim trailing spaces
        choice = input("Select an option (1-4): ").strip()

        # Handle user's menu choice using conditional branching
        if choice == "1":
            check_balance(balance)
        elif choice == "2":
            balance = deposit_money(balance)
        elif choice == "3":
            balance = withdraw_money(balance)
        elif choice == "4":
            print("\nThank you for using our ATM. Goodbye!")
            break  # Exit the infinite while loop
        else:
            print("Invalid selection. Please choose a valid menu option (1-4).")


# Entry point of the script: run the ATM simulation when executed directly
if __name__ == "__main__":
    atm_simulation()
```
