```python
# Print numbers from 1 to 100 using while.

# Initialization: Set the starting value of the counter variable to 1
i = 1

# Condition: Continue looping as long as i is less than or equal to 100
while i <= 100:
    # Print the current value of i
    print(i)

    # Update: Increment the counter by 1 in each iteration to avoid an infinite loop
    i += 1
```

```python
# Print numbers from 100 to 1 using while loop

# Initialization: Start the counter variable at 100
i = 100

# Condition: Continue looping as long as i is greater than or equal to 1
while i >= 1:
    # Print the current value of i
    print(i)

    # Update/Termination: Decrement the counter by 1 in each iteration to reach the exit condition
    i -= 1
```

```python
# Print all even numbers between 1 and 100

# Initialization: Start counter variable at 2 (the first even number)
i = 2

# Condition: Continue looping as long as i is less than or equal to 100
while i <= 100:
    # Check if the current value of i is divisible by 2
    if i % 2 == 0:
        # Print the even number along with a label
        print(i, "even number")

    # Update/Termination: Increment the counter by 1 in each iteration
    i += 1
```

```python
#Calculate the sum of digits of a number.
# Initialization: Get user input and setup sum variable
number = int(input("Enter a number: "))
sum = 0

# Condition: Continue until all digits are extracted (number becomes 0)
while number > 0:
    # Extract: Get the last digit using modulus operator
    digit = number % 10

    # Accumulate: Add the extracted digit to sum
    sum += digit

    # Update/Termination step: Remove the last digit using integer division
    number //= 10

# Print final result
print("The sum of digits is:", sum)
```

Loop Exit

Loop stops. Final sum = 18.

```python
# Reverse an integer using a while loop.

# Initialization: Take an integer input from the user and initialize the reversed number to 0
number = int(input("Enter a number: "))
reversed_number = 0

# Condition: Continue processing digits as long as the remaining number is greater than 0
while number > 0:
    # Extract the last digit of the current number
    digit = number % 10
    
    # Append the extracted digit to the right side of reversed_number
    reversed_number = reversed_number * 10 + digit
    
    # Update/Termination: Perform integer division by 10 to remove the last processed digit
    number //= 10

# Print the final reversed integer
print("The reversed number is:", reversed_number)
```

How the Logic Works:

digit = number % 10: Extracts the last digit of number.
reversed_number = reversed_number * 10 + digit: Shifts the current digits in reversed_number one place to the left (by multiplying by 10) and appends the extracted digit at the end.
number //= 10: Removes the last digit from number by performing integer division.

```python
# Count the number of digits in an integer.

# Initialization: Prompt the user for an integer and initialize the count tracker to 0
number = int(input("Enter a number: "))
count = 0

# Condition: Continue looping as long as there are remaining digits in the number
while number > 0:
    # Update/Termination: Remove the last digit from the number using integer division
    number //= 10
    
    # Increment the digit count by 1 for each digit removed
    count += 1

# Display the total count of digits
print("The number of digits is:", count)
```

How the Logic Works
number //= 10: Performs integer division to remove the last digit from number.

count += 1: Increments the digit counter by 1 for every digit removed.

while number > 0: Keeps running as long as there are digits remaining in number.

```python
# Calculate factorial using while.


# Initialization: Get user input, initialize factorial result to 1, and set counter
number = int(input("Enter a number: "))
factorial = 1
i = number

# Condition: Continue as long as counter is greater than 0
while i > 0:
    # Accumulate: Multiply factorial by current value of i
    factorial *= i
    
    # Update/Termination step: Decrement i by 1
    i -= 1

# Output final result
print(f"The factorial of {number} is:", factorial)
```

```python
# Create a program that repeatedly asks for numbers and stops only when the user enters 0. 
# Display the sum of all previously entered numbers.

# Initialization: Variable to hold the running total, and initial user prompt
total_sum = 0
number = int(input("Enter a number (0 to stop): "))

# Condition: Keep looping as long as the user hasn't entered 0
while number != 0:
    # Accumulate: Add the entered number to the total sum
    total_sum += number
    
    # Update/Termination step: Ask for the next number
    number = int(input("Enter a number (0 to stop): "))

# Output final result
print("The sum of all entered numbers is:", total_sum)
```

```python
# Create a password checker that keeps asking for the password until the correct password is entered.

# Initialization: Store the expected password and take the initial password input from the user
correct_password = "Python123"
password = input("Enter the password: ")

# Condition: Continue prompting as long as the user's input does not match the correct password
while password != correct_password:
    print("Incorrect password. Please try again.")
    
    # Update/Termination: Re-prompt for input; entering the correct password updates the condition to False
    password = input("Enter the password: ")

# Final execution step after exiting the loop
print("Password accepted.")
```

```python
# Create a guessing game where the secret number is predefined and the user keeps guessing until correct.

# Initialization: Predefine the secret number and prompt the user for their first guess
secret_number = 7
guess = int(input("Guess the number (1-10): "))

# Condition: Continue looping as long as the user's guess does not match the secret number
while guess != secret_number:
    print("Wrong guess. Try again.")
    
    # Update/Termination: Prompt for a new guess; matching secret_number breaks the condition
    guess = int(input("Guess the number (1-10): "))

# Final execution step after exiting the loop
print("Congratulations! You guessed the number.")
```

```python
# Build a menu-driven calculator. The menu should continue until the user chooses Exit.
# Initialization: Variable to track user choice
choice = 0

# Condition: Continue looping as long as the choice is not 5 (Exit)
while choice != 5:
    # Display the menu options
    print("\n--- Calculator Menu ---")
    print("1. Add")
    print("2. Subtract")
    print("3. Multiply")
    print("4. Divide")
    print("5. Exit")
    
    # Update/Termination step (1): Get user choice
    choice = int(input("Enter your choice (1-5): "))
    
    # Check if user wants to exit
    if choice == 5:
        print("Exiting calculator. Goodbye!")
    elif choice in (1, 2, 3, 4):
        # Prompt for numbers only when a valid arithmetic operation is selected
        num1 = float(input("Enter first number: "))
        num2 = float(input("Enter second number: "))
        
        # Perform operation based on choice
        if choice == 1:
            print(f"Result: {num1} + {num2} = {num1 + num2}")
        elif choice == 2:
            print(f"Result: {num1} - {num2} = {num1 - num2}")
        elif choice == 3:
            print(f"Result: {num1} * {num2} = {num1 * num2}")
        elif choice == 4:
            # Check for division by zero
            if num2 != 0:
                print(f"Result: {num1} / {num2} = {num1 / num2}")
            else:
                print("Error: Division by zero is not allowed.")
    else:
        print("Invalid choice! Please select a valid option from 1 to 5.")
```

```python
# Create an ATM menu using while where the application continues running until the user explicitly chooses Exit.

# Initialization: Initial balance and choice variable
balance = 1000.0  # Starting balance
choice = 0

# Condition: Continue running as long as the user does not choose option 4 (Exit)
while choice != 4:
    # Display the ATM Menu
    print("\n--- ATM Menu ---")
    print("1. Check Balance")
    print("2. Deposit Money")
    print("3. Withdraw Money")
    print("4. Exit")

    # Update/Termination step (1): Get user choice
    choice = int(input("Choose an option (1-4): "))

    # Process user choice
    if choice == 1:
        # Display the account balance formatted to two decimal places
        print(f"Your current balance is: ${balance:.2f}")

    elif choice == 2:
        # Prompt user to input deposit amount
        amount = float(input("Enter amount to deposit: $"))

        # Verify deposit amount is positive before updating balance
        if amount > 0:
            balance += amount
            print(f"Successfully deposited ${amount:.2f}. New balance: ${balance:.2f}")
        else:
            print("Invalid deposit amount!")

    elif choice == 3:
        # Prompt user to input withdrawal amount
        amount = float(input("Enter amount to withdraw: $"))

        # Validate that funds are sufficient and the amount is positive
        if amount > balance:
            print("Insufficient funds! Transaction cancelled.")
        elif amount <= 0:
            print("Invalid withdrawal amount!")
        else:
            # Deduct the withdrawal amount from the balance
            balance -= amount
            print(f"Successfully withdrew ${amount:.2f}. New balance: ${balance:.2f}")

    elif choice == 4:
        # Exit execution step triggered when choice is 4
        print("Thank you for using the ATM. Goodbye!")

    else:
        # Catch-all for choices outside the 1–4 range
        print("Invalid selection! Please choose an option between 1 and 4.")
```
