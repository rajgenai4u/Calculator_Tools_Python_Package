#Explain how indexing works with nested lists.

In a nested list, each element is itself a list. To access an element in a nested list, you use two indices: the first index refers to the position of the inner list, and the second index refers to the position of the element within that inner list.
for example, in the list students, students[0] refers to the first student's record (a list), and students[0][0] refers to the 
first student's name (the first element of that inner list).
