```python
#String creation
message = " Welcome To Python Programming Class "
print(message)
```

```python
#Remove extra spaces
message = message.strip()
print(message)
```

```python
#Convert everything to lowercase
message = message.lower()
print(message)
```

```python
#Convert everything to uppercase
message = message.upper()
print(message)
```

```python
#Convert to title case
message = message.title()
print(message)
```

```python
#Replace "Python" with "Advanced Python"
message = message.replace("Python", "Advanced Python")
print(message)
```

```python
#Check whether the string starts with "Welcome"
message1= message.startswith("Welcome")
print(message1)
```

```python
#Check whether it ends with "Class"
message2= message.endswith("Class")
print(message2)
```

```python
#Count occurrences of "o"
message.count("o")
```

```python
#Find the position of "Programming"
position = message.find("Programming")
print(position)
```

```python
#Split the sentence into words
message.split()
```

# Students must explain the difference between the below functions

strip() (Whitespace & Character Removal): Removes leading and trailing whitespace (or specific characters passed as arguments) from both ends of the string. It does not modify spaces in the middle.

lower() (Case Conversion): Converts all uppercase characters in the string to lowercase.

upper() (Case Conversion): Converts all lowercase characters in the string to uppercase.

title() (Case Formatting): Converts the first character of each word to uppercase and all other letters to lowercase.

replace(old, new) (String Substitution): Replaces all occurrences of a specified substring (old) with a new substring (new).

find(sub) (Position Lookup): Returns the lowest index (0-based) where the substring sub is found in the string. Returns -1 if the substring is not found.

count(sub) (Frequency Counting): Returns the total number of non-overlapping occurrences of the substring sub in the string.

split(sep) (String Parsing): Breaks up a string into a list of substrings using a delimiter (sep). If no delimiter is specified, it splits by any whitespace.
