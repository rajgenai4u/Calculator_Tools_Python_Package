```python
 #Create Tuple
technologies = ("Python", "Java", "Python", "C++", "JavaScript", "Python")
```

```python
#Print Tuple & type

print(technologies)

print(type(technologies))
```

```python
#Print the first item
print(technologies[0])
```

```python
#Print the last item
print(technologies[-1])
```

```python
#slice the touple
print(technologies[2:5])

print(technologies[::2]) # print every 2nd element
```

```python
#Count "Python"
print(technologies.count("Python"))
```

```python
#Find the index of "C++"
print(technologies.index("C++"))
```

```python
#Find the tuple length
print(len(technologies))
```

```python
#Convert the tuple into a list
technologies=list(technologies)
print(technologies)
print(type(technologies))
```

```python
#Add "Go" after converting it into a list
technologies.append("Go")
print(technologies)
```

```python
#Convert it back into a tuple
technologies=tuple(technologies)
print(technologies)
print(type(technologies))
```

Explain why tuples are called immutable:

Tuples are called immutable because once a tuple is created in memory, its size and the element references it contains cannot be modified, added, removed, or reassigned.

```python
## Consolidated Code

# Create the initial tuple
technologies = ("Python", "Java", "Python", "C++", "JavaScript", "Python")

# 1. Print the tuple
print("Tuple:", technologies)

# 2. Print its type
print("Type:", type(technologies))

# 3. Print the first item
print("First item:", technologies[0])

# 4. Print the last item
print("Last item:", technologies[-1])

# 5. Slice the tuple (e.g., items from index 1 to 3)
print("Sliced tuple (index 1 to 3):", technologies[1:4])

# 6. Count "Python"
python_count = technologies.count("Python")
print("Count of 'Python':", python_count)

# 7. Find the index of "C++"
cpp_index = technologies.index("C++")
print("Index of 'C++':", cpp_index)

# 8. Find the tuple length
print("Tuple length:", len(technologies))

# 9. Convert the tuple into a list
tech_list = list(technologies)

# 10. Add "Go" after converting it into a list
tech_list.append("Go")
print("Updated list after adding 'Go':", tech_list)

# 11. Convert it back into a tuple
technologies_updated = tuple(tech_list)
print("Updated tuple:", technologies_updated)
```
