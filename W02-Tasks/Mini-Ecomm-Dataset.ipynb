```python
#Create a Mini E-Commerce Dataset

products = [
    {
        "name": "Headphones",
        "price": 15000,
        "brand": "Sony"
    },
    {
        "name": "Smartwatch",
        "price": 25000,
        "brand": "Garmin"
    },
    {
        "name": "Monitor",
        "price": 35000,
        "brand": "LG"
    }
]
```

```python
#Print all products without using a loop
print(products)
```

```python
#Print second product's price
print(products[1]["price"])
```

```python
#Change first product's price
products[0]["price"] = 45000
print(products[0])
```

# Explain list() Dictionary() Index() Keys() and values()

List: An ordered collection that holds multiple data items (like dictionaries) sequentially using zero-based position index numbers.

Dictionary: An unordered structure that stores data as key-value pairs representing individual attributes of an object.

Indexing: The method of specifying an index position (e.g., products[0]) to select a specific dictionary item out of a list.

Keys: The unique identifier strings inside a dictionary used to look up specific properties (e.g., 'price').

Values: The actual data stored within a dictionary that corresponds to a specific key (e.g., 70000).
