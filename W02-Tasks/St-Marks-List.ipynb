```python
# Create List with name Marks

marks = [78, 85, 90, 67, 88, 92, 76]
```

```python
# Print the complete list
print("Marks list", marks)
```

```python
#Print the complete list
marks
```

```python
#Print the first element
print("First Element in the list is:", marks[0])
```

```python
#Print the last element
print("Last element in the list is:", marks[-1])
```

```python
#Print elements from index 2 to 5
print("Elements from Index 2 to 5 are :", marks[2:6])
```

```python
#Find the number of elements
print("number of elements in the list are:", len(marks))
```

```python
#Find maximum marks
print("Maximum marks in the list:", max(marks))
```

```python
#Find minimum marks
print("Minimum marks in the list :", min(marks))
```

```python
#Find total marks
print("Total marks in the list are:", sum(marks))
```

```python
#Sort marks in ascending order
print("Marks in ascending order:", sorted(marks))
```

```python
#Sort marks in descending order
print("Marks in descending order:", sorted(marks, reverse=True))
```

```python
#Add 95
marks.append(95)
print(marks)
```

```python
#Add [81, 84]
marks.extend([81,84])
marks
```

```python
#Remove 95
marks.remove(95)
marks
```

```python
#Count how many times 95 occurs
print("Number of times 95 occurs in the list:" ,marks.count(95))
```

```python
#Find the index of 88
print("Index of 88 is:", marks.index(88))
```
