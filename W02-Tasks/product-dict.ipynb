```python
#Create a dictionary representing a laptop
laptop = {

"brand": "Dell",

"model": "XPS 15",

"price": 120000,

"ram": "16GB",

"storage": "512GB SSD",

"available": True

}
```

```python
# Display the product
print(laptop)
print(type(laptop))
```

```python
#Print brand
print(laptop["brand"])
```

```python
#Print model
print(laptop["model"])
```

```python
#Print price
print(laptop["price"])
```

```python
#change price
laptop["price"] = 150000
print(laptop["price"])
```

```python
#Add processor information
laptop["processor"] = "Intel Core i7"
print(laptop)
```

```python
#Add processor information
laptop["GPU"] = "NVIDIA GTX 1650"
print(laptop)
```

```python
#Change RAM to "32GB"
laptop["ram"]= "32GB"
print(laptop)
```

```python
#Remove "available"
laptop.pop("available")
print(laptop)
```

```python
#Print all keys
print(laptop.keys())
```

```python
#Print all values
print(laptop.values())
```

```python
#Print all items
print(laptop.items())
```
