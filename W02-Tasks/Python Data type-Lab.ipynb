```python
# ==========================================
# PART 1: 15 Variables & Type Identification
# ==========================================

# Integers (int)
age = 25
year = 2026

# Floating-point numbers (float)
price = 785.20
avg= 85.6

# Strings (str)
user_name = "Rajesh"
city = 'Hyderabad'

# Booleans (bool)
is_active = True
has_discount = False

# Lists (list)
num= [85, 90, 95]
fruits = ["apple", "banana", "cherry"]

# Tuples tuple)
coordinates = (10.0, 20.0)
rgb= (255, 128, 0)

# Sets (set)
ids = {101, 102, 103}
tags = {"python", "coding"}

# Dictionaries (dict)
student_info = {"name": "Rajesh", "grade": "A"}


```

```python
# Displaying each type using type() 
print("--- VARIABLE TYPES ---")
print(age)
print("age", type(age))
print("year:", type(year))
print(price)
print("price:", type(price))
print("avg:", type(avg))
print(user_name)
print("user_name:", type(user_name))
print("city:", type(city))
print(is_active)
print("is_active:", type(is_active))
print("has_discount:", type(has_discount))
print(num)  
print("num:", type(num))
print("fruits:", type(fruits))
print("coordinates:", type(coordinates))
print(rgb)  
print("rgb:", type(rgb))
print("ids:", type(ids))
print(tags)
print("tags:", type(tags))
print(student_info)
print("student_info:", type(student_info))
```

```python


# ==========================================
# PART 2: Basic Type Conversions & Explanations
# ==========================================
print("\n--- TYPE CONVERSIONS ---")

# 1. String to Integer
conv1 = int("100")
print(conv1)
print(type(conv1))
# Explanation: converts the text string "100" and returns the integer value 100, allowing mathematical operations.
print("\n")

# 2. String to Float
conv2 = float("45.67")
print(conv2)
print(type(conv2))
# Explanation: Converts the decimal string  "45.67" into a floating point number 45.67.
print("\n")

# 3. Integer to String
conv3 = str(500)
print(conv3)
print(type(conv3))
# Explanation: Converts the integer numeric value 500 into a sequence of characters "500".
print("\n")

# 4. Integer to Boolean
conv4 = bool(1)
print(conv4)
print(type(conv4))
# Explanation: Evaluates truthiness; non-zero numbers evaluate to True (0 evaluates to False).
print("\n")

# 5. Tuple to List
conv5 = list((1, 2, 3))
print(conv5)
print(type(conv5))
# Explanation: Converts the immutable tuple (1, 2, 3) into a mutable list [1, 2, 3].
print("\n")

# 6. List to Tuple
conv6 = tuple([1, 2, 3])
print(conv6)
print(type(conv6))
# Explanation: Converts the mutable list [1, 2, 3] into an immutable tuple (1, 2, 3).
print("\n")

# 7. List to Set
conv7 = set([1, 2, 2, 3])
print(conv7)
print(type(conv7))
# Explanation: Converts the list to an unordered set {1, 2, 3} and automatically removes duplicate elements (the extra 2).
print("\n")
```
