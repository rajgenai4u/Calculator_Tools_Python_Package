```python
#Create a list
numbers = [10, 20, 30, 20, 40, 10, 50, 30, 60]
```

```python
# Print the original list
print(numbers)
```

```python
#Convert the list into a set
numset= set(numbers)
print(numset)
```

```python
#Observe which duplicates disappear

# duplicated disappered - 10,20,30
```

```python
#Convert the set back into a list
numlist=list(numset)
print(numlist)
```

```python
#Print the number of original elements
print(len(numbers))
```

```python
#Print the number of unique elements
print(len(numlist))
```

```python
# Create another example using duplicate student names.
# Create a list with duplicate student names
student_names = ["Sudhanshu", "Rajesh", "Ramu", "Rajesh", "Jitendra", "Ramu", "Sravan", "Jitendra"]

# Print the original list
print("Original student list:", student_names)

# Convert the list into a set to remove duplicates
unique_names_set = set(student_names)

# Convert the set back into a list
unique_names = list(unique_names_set)

# Output results
print("Unique student list:", unique_names)
print("Number of original elements:", len(student_names))
print("Number of unique elements:", len(unique_names))
```

##Explain why sets are useful when working with duplicate data.

1. Automatically remove duplication: ets are unordered collections of unique elements. When you convert a list or collection containing duplicate items into a set, Python automatically eliminates all duplicate
2. Efficient Set Operations: Sets provide native, optimized methods for mathematical set operations:
Union (|): Combines distinct elements from multiple datasets.

Intersection (&): Finds common elements across datasets while automatically ignoring duplicates.

Difference (-): Identifies unique items present in one dataset but absent in another.
