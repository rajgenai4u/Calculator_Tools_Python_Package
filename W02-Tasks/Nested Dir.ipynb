```python
# Consolidated Code

# 1. Create the employee dictionary
employee = {
    "name": "Amit",
    "department": "Engineering",
    "skills": {
        "language": "Python",
        "database": "PostgreSQL",
        "cloud": "AWS"
    },
    "salary": 80000
}

# 2. Print employee name
print("Employee Name:", employee["name"])

# 3. Print department
print("Department:", employee["department"])

# 4. Print complete skills dictionary
print("Complete Skills Dictionary:", employee["skills"])

# 5. Print programming language
print("Programming Language:", employee["skills"]["language"])

# 6. Print database
print("Database:", employee["skills"]["database"])

# 7. Print cloud technology
print("Cloud Technology:", employee["skills"]["cloud"])

# 8. Change "Python" to "Python + JavaScript"
employee["skills"]["language"] = "Python + JavaScript"
print("Updated Language:", employee["skills"]["language"])

# 9. Change salary
employee["salary"] = 95000
print("Updated Salary:", employee["salary"])

# 10. Add "experience": 3
employee["experience"] = 3
print("Added Experience:", employee["experience"])

# 11. Add another skill under the skills dictionary
employee["skills"]["tool"] = "Docker"
print("Updated Skills Dictionary:", employee["skills"])

# Print the entire updated employee structure
print("\nFinal Employee Object:")
print(employee)
```
