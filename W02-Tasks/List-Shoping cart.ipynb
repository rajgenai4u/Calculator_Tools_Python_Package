# Explain the difference between
append()
extend()
insert()
remove()
pop()
clear()
copy()
sort()
reverse()



append(item): Adds a single element to the end of the list and modifies the original list in-place.

extend(iterable): Appends every element from an iterable (e.g., another list) to the end of the current list in-place.

insert(index, item): Places an element at a specified index position, shifting existing items to the right.

remove(item): Deletes the first occurrence of a specified value from the list in-place.

pop([index]): Removes and returns the item at a given index (defaults to the last item if no index is provided).

clear(): Deletes all elements from the list, leaving it completely empty.

copy(): Creates and returns a shallow copy of the list without modifying the original list.

sort(): Rearranges all items in ascending order directly inside the original list.

reverse(): Flips the order of elements in the list in-place.
