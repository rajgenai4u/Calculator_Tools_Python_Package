```python
# create sets

python_students = {"Rahul", "Aman", "Priya", "Karan", "Neha"}
java_students = {"Priya", "Karan", "Rohit", "Simran"}
```

```python
#Print both sets

print(python_students)
print(java_students)
```

```python
#Find students learning either Python or Java

students=python_students.union(java_students)

#students = python_students | java_students

print("students learning either Python or Java are: ", students)
```

```python
#Find students learning both

students1= python_students.intersection(java_students)
print(" Students learning both Python & Java are: ", students1)
```

```python
#Find students learning only Python

Pystudents= python_students.difference(java_students)
print(Pystudents)
```

```python
#Find students learning only Java

Javastudents_only= java_students.difference(python_students)
print(Javastudents_only)
```

```python
#Find students belonging to exactly one group

stud=python_students.symmetric_difference(java_students)
print("students belonging to exactly one group are :",stud)
```

```python
#Add a new student
python_students.add("Rajesh")

# Print the updated set
print("Updated Python Students:", python_students)
```

```python
#Remove a student
python_students.remove("Rajesh")

# Print the updated set
print("Updated Python Students:", python_students)
```

```python
#Remove an existing student using discard()
python_students.discard("Neha")
print("After discarding Neha:", python_students)
```

```python
# Attempt to remove a student who is NOT in the set
python_students.discard("Raj")
print("After discarding non-existent student Raj", python_students)
```

```python
# Attempt to remove a student who is NOT in the set
python_students.remove("Raj")
print("After discarding non-existent student Raj", python_students)
```

Key Difference: discard() vs remove()

discard() (Safe Removal): Removes the specified element if present. If the element is missing, it does nothing and raises no errors.

remove() (Strict Removal): Removes the specified element if present, but raises a KeyError if the element does not exist in the set.

# Initial sets of students
python_students = {"Rahul", "Aman", "Priya", "Karan", "Neha"}
java_students = {"Priya", "Karan", "Rohit", "Simran"}

all_students = python_students.union(java_students)

both_courses = python_students.intersection(java_students)
print("Intersection:", both_courses)

# 3. difference() - Students learning only Python
only_python = python_students.difference(java_students)
print("Difference (Only Python):", only_python)

# 4. symmetric_difference() - Students in exactly one group
only_one_course = python_students.symmetric_difference(java_students)
print("Symmetric Difference:", only_one_course)

# 5. add() - Adding a new student
python_students.add("Vikram")
print("After add():", python_students)

# 6. remove() - Removing a student (raises error if missing)
python_students.remove("Aman")
print("After remove():", python_students)

# 7. discard() - Safely removing a student (no error if missing)
python_students.discard("NonExistentStudent")
print("After discard():", python_students)
