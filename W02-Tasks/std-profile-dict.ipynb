# Explain  keys() values()  items()  get() update() pop() copy()

keys(): Returns a view object containing all the keys present in the dictionary.

values(): Returns a view object containing all the values stored in the dictionary.

items(): Returns a view object containing key-value pairs as tuples ((key, value)).

get(): Retrieves the value for a specified key, returning None or a custom default if the key is not found (avoiding key errors).

update(): Merges another dictionary or key-value pairs into the existing dictionary, adding new keys and overwriting existing ones.

pop(): Removes the specified key from the dictionary and returns its corresponding value (or a default if provided).

copy(): Creates and returns a shallow copy of the dictionary

n Python, dictionaries do not have a built-in method named merge(). Merging dictionaries is typically done using either the update() method or the merge operator | (introduced in Python 3.9).

update() is a built-in method for Python dictionaries, while merge() is the one that doesn't exist as a built-in method.


1. Using update() (Modifies in-place) :
---------------------------------------

The update() method modifies the original dictionary by inserting key-value pairs from another dictionary, overwriting existing keys.

dict1 = {"a": 1, "b": 2}
dict2 = {"b": 99, "c": 4}

dict1.update(dict2)

print(dict1)  # Output: {'a': 1, 'b': 99, 'c': 4}

2. Using the Merge Operator | (Creates a new dictionary) :
----------------------------------------------------------

If you are looking for a functional "merge" operation that leaves the original dictionaries unchanged and returns a brand-new combined dictionary, use the | operator:

dict1 = {"a": 1, "b": 2}
dict2 = {"b": 99, "c": 4}

merged_dict = dict1 | dict2

print(merged_dict)  # Output: {'a': 1, 'b': 99, 'c': 4}
print(dict1)        # Output: {'a': 1, 'b': 2} (original remains unchanged)

```python
dict1 = {"a": 1, "b": 2}
dict2 = {"b": 99, "c": 4}

dict1.update(dict2)

print(dict1)
```

```python
dict1 = {"a": 1, "b": 2}
dict2 = {"b": 99, "c": 4}

merged_dict = dict1 | dict2

print(merged_dict)  # Output: {'a': 1, 'b': 99, 'c': 4}
print(dict1)        # Output: {'a': 1, 'b': 2} (original remains unchanged)
```
